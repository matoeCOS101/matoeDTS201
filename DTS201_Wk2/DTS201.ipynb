```python
import pandas as pd

# 1. Load the Excl file
df = pd.read_excel("50_startups_messy.xlsx")

print("Original Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())


# 2. Clean the data

# Remove duplicate rows
df = df.drop_duplicates()

# Standardise State names
df["State"] = df["State"].str.strip().str.title()

# Convert Marketing Spend to numeric
df["Marketing Spend"] = pd.to_numeric(
    df["Marketing Spend"], errors="coerce"
)

print("\nCleaned Data:")
print(df.head())

print("\nCleaned Shape:", df.shape)

print("\nState Values:")
print(df["State"].value_counts())

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())


# 3. Create a simple random sample of 20
simple_sample = df.sample(n=20, random_state=7)

print("\nSimple Random Sample:")
print(simple_sample)

print("\nSimple Sample Size:", len(simple_sample))


# Create a stratified 20% sample by State
stratified_sample = (
    df.groupby("State", group_keys=False)
      .sample(frac=0.20, random_state=7)
)

print("\nStratified Sample:")
print(stratified_sample)

print("\nStratified Sample Size:", len(stratified_sample))

print("\nNumber from Each State:")
print(stratified_sample["State"].value_counts())


# 4. Compare the means

# Population means
population_mean = df[["R&D Spend", "Marketing Spend"]].mean()

# Simple sample means
simple_mean = simple_sample[["R&D Spend", "Marketing Spend"]].mean()

# Stratified sample means
stratified_mean = stratified_sample[
    ["R&D Spend", "Marketing Spend"]
].mean()


print("\nPopulation Means:")
print(population_mean)

print("\nSimple Random Sample Means:")
print(simple_mean)

print("\nStratified Sample Means:")
print(stratified_mean)


# Calculate the difference from the population
simple_difference = (simple_mean - population_mean).abs()
stratified_difference = (stratified_mean - population_mean).abs()

print("\nDifference from Population - Simple Sample:")
print(simple_difference)

print("\nDifference from Population - Stratified Sample:")
print(stratified_difference)
```
